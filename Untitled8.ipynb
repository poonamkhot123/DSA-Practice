```python
# Two sum II - input array is sorted.
def twoSum(nums,target):
  left = 0
  right = len(nums) - 1

  while left<right:
    sum1 = nums[left] + nums[right]
    if sum1 == target:
      return [left,right]
    elif sum1>target:
      right-=1
    else:
      left+=1
```

```python
nums = [3,4,5,6,7,8,9]
twoSum(nums,12)
```

```python
def intersection(nums1,nums2):
  set1 = set(nums1)
  set2 = set(nums2)

  return set1.intersection(set2)
```

```python
nums1 = [3,4,5,6,7]
nums2 = [6,7,8,9,3]
intersection(nums1,nums2)
```

```python
# first unique character in a string.
def firstUniqueChar(s):
  freq = {}
  for i in s:
    if i not in freq:
      freq[i]  = 1
    else:
      freq[i] += 1
  for i in range(len(s)):
    if freq[s[i]] == 1:
      return i
  return -1
```

```python
# anagram string
def isAnagram(s,t):
  if len(s) != len(t):
    return False
  freq = {}
  for i in s:
    if i not in freq:
      freq[i] = 1
    else:
      freq[i] += 1
  for i in t:
    if i not in freq:
      return False
    else:
      freq[i] -= 1
  for i in freq.values():
    if i!=0:
      return False
  return True
```

```python
s = "anagram"
t = "nagaram"
isAnagram(s,t)
```

```python
# Group Anagram
def sortString(s):
  s1 = list(s)
  s1.sort()
  return "".join(s1)

def groupAnagrams(strs):
  dict1 = {}

  for s in strs:
    key = sortString(s)
    if key in dict1:
      dict1[key].append(s)
    else:
      dict1[key] = [s]
  return list(dict1.values())
```

```python
s = ['eat','tea','tan','ate','nat','bat']
groupAnagrams(s)
```

```python
s = "aababcabc"
def countSubstrings(s):
  n = len(s)
  ans = 0

  for i in range(n-2):
    if s[i] != s[i+1] and s[i+1] != s[i+2] and s[i+2] != s[i]:
      ans += 1
  return ans
```

```python
countSubstrings(s)
```

```python
# Longest substring with distinct characters.
def length(s):
  n = len(s)
  if n== 0:
    return 0

  ans = 1
  set1 = set({})
  set1.add(s[0])

  i = 0
  j = 1

  while j < n:
    while s[j] in set1:
      set1.discard(s[i])
    set1.add(s[j])
    j+=1
    ans = max(ans,(j-i))
  return ans
```
