```python
# Recursion
n =5
i =1

while i<=n:
  print(i)
  i+=1
```

```python
def printNumbers(i,n):
  # base case
  if i>n:
    return
  # recursive case
  print(i)
  printNumbers(i+1,n)
```

```python
printNumbers(1,10)
```

```python
# factorial of number
def fact(n):
  if n==0:
    return 1
  return n*fact(n-1)
```

```python
fact(5)
```

```python
# fibonacci Number
def fib(n):
  if n == 0 or n == 1:
    return n
  return fib(n-1) + fib(n-2)
```

```python
fib(6)
```

```python
# Power of Two
def isPowerofTwo(n):
  if n <=0:
    return False
  if n== 1:
    return True
  if n%2 !=0 :
    return False
  return isPowerofTwo(n//2)
```

```python
isPowerofTwo(16)
```

```python
# GCD of Two Numbers
def gcd(a,b):
  if b == 0:
    return a
  return gcd(b,a%b)
```

```python
# pov(x,n)
def findPow(x,n):
  if n == 0:
    return 1
    a = findPow(x,n//2)
    if n %2==0:
      return a*a
    else:
      return a*a*x

def myPow(x,n):
  if n>=0:
    return findPow(x,n)
  else:
    return 1/findPow(x,-n)
```

```python
myPow(2,3)
```
