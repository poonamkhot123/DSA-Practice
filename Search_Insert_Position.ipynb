```python
# search insert position.
def lowerBound(nums,target):
  n = len(nums)
  l = 0
  r = n-1
  ans = n

  while l <=r:
    mid = (l+r)//2

    if nums[mid] >= target:
      ans = mid
      r = mid-1
    else:
      # right
      l = mid+1
  return ans

def searchInsert(nums,target):
  return lowerBound(nums, target)
```

```python
nums = [4,5,6,7,8,9,12]
searchInsert(nums,10)
```
