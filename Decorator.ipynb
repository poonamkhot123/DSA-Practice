```python
def decorator(function):
  def wrapper():
    print("Before function calling")

    function()

    print("After function calling")
  return wrapper

def sayHello():
  print("I am learning decorators...")

new_function = decorator(sayHello)
new_function()
```

```python
def akash_decorator(function):
  def wrapper():
    print("Before function calling")

    function()

    print("After function calling")
  return wrapper

@akash_decorator
def greet():
  print("hello")
greet()

```
