```python
import math
# By using Binary Search
def searchmatrix(matrix,target):
  rows = len(matrix)
  cols = len(matrix[0])

  l = 0
  r = rows*cols - 1

  while l <=r:
    mid = (l+r)//2

    i = mid//cols
    j = mid % cols

    if matrix[i][j] == target:
      return True
    elif matrix[i][j] > target:
      # left
      r = mid -1
    else:
      l = mid +1
  return False
```

```python
matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]]
searchmatrix(matrix,34)
```
