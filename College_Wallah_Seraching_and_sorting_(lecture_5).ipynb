```python
# Bubble sort
def sortArray(nums):
  n = len(nums)

  for i in range(n):
    isSwap = False
    for j in range(n-i-1):
      if nums[j] > nums[j+1]:
        # swap
        temp = nums[j]
        nums[j+1] = temp
        isSwap = True
    if not isSwap:
      break
```

```python
# Insertion sort
def sortArray(nums):
  n = len(nums)

  for i in range(1,n):
    key = nums[i]
    j = i -1
    while j>=0 and nums[j] > key:
      nums[j+1] = nums[j]
      j-=1
    nums[j+1] = key

  return nums
```

```python
# selection sort
def sortArray(nums):
  n = len(nums)

  for i in range(n):
    mn = min(mn,nums[j])
    ind = i
    for j in range(i+1,n):
      if nums[j]<mn:
        mn = nums[j]
        ind = j

    temp = nums[i]
    nums[i] = nums[ind]
    nums[ind] = temp

  return nums
```

```python
# Quick sort
def partition(nums,l,r):
  key = nums[r]
  start = l

  for i in range(l,r+1):
    if nums[i]<=key:
      temp = nums[i]
      nums[i] = nums[start]
      nums[start] = temp
      start += 1
  return start - 1


def quickSort(nums,l,r):
  # base case
  if l>=r:
    return
  p =partition(nums,l,r)
  quickSort(nums,l,p-1)
  quickSort(nums,p+1,r)

def sortArray(nums):
  n = len(nums)
  quickSort(nums,0,n-1)
  return nums
```

```python
# counting Sort
def sortArray(nums):
  n = len(nums)
  maxi = max(nums)

  freq = [0] * (maxi+1)

  for i in nums:
    freq[i] +=1

  for i in range(0,maxi+1):
    while freq[i]>0:
      nums.append(i)
      freq[i]-=1
  return nums

```
