```python
# count Odd Numbers in an Interval Range.
def countOdds(self,low,high):
  return (high+1)//2 - (low//2)
```

```python
countOdds(1,10,99)
```

```python
# How many Numbers are smaller than the current Number.
class Solution:
  def smallerNumber(self,nums):
    ans = []
    for i in nums:
      c = 0
      for j in nums:
        if j<i:
          c+=1
      ans.append(c)
    return ans
```

```python
# Count the digits that Divide a Number.
class Solution:
  def countDigits(self,num):
    temp = num
    ans = 0
    while temp >0:
      r = temp % 10
      if num %r == 0:
        ans += 1
      temp //=10

    return ans
```

```python
solution_instance = Solution()
print(solution_instance.countDigits(1248))
```

```python
# Palindrome number.
def palindrome(n):
    nums = n
    total = 0
    while nums>0:
      ld = nums % 10
      total = total * 10 + ld
      nums = nums // 10
    return total == n
```

```python
palindrome(121)
```

```python
# Subtract the Product and Sum of Digits of an Integer.
def Productandsum(n):
  temp =n
  addition = 0
  product = 1
  while temp >0:
    r = temp %10
    temp = temp // 10
    addition = addition + r
    product = product * r
  return product - addition
```

```python
# kids with the Greatest Number of Cnadies.
candies = [2,3,5,1,3]
extracandies = 3
def kidswithcandies(candies,extracandies):
  maxCandies = max(candies)
  ans = []

  for i in candies:
    if (i+extracandies) >= maxCandies:
      ans.append(True)
    else:
      ans.append(False)
  return ans
```

```python
kidswithcandies(candies,6)
```

```python
# by using list comprehension.
def kidswithcandies(candies,extracandies):
  return [(i+extracandies)>=max(candies) for i in candies]
```

```python
kidswithcandies(candies,3)
```
