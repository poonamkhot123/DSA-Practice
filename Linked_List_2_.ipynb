```python
class Node:
  def __init__(self,data):
    self.data = data
    self.next = None

a = Node(5)
b = Node(3)
c = Node(7)

a.next = b
b.next = c

head = a
print(head.data)
```

```python
# Traverse Linked list

def printLinkedList(head):
  curr = head
  while curr != None:
   print(curr.data,end = " ")
   curr = curr.next
```

```python
# insertion at the beggining
newNode = Node(4)
newNode.next = head
head = newNode

printLinkedList(head)
```

```python
# insertion at the end.
newNode = Node(1)
curr = head
newNode.next = None
while curr.next != None:
  curr = curr.next

curr.next = newNode


printLinkedList(head)
```

```python
#  insertion at the kth index
k = 2
newNode = Node(6)
curr = head
for i in range(k-1):
  curr = curr.next

newNode.next = curr.next
curr.next = newNode

printLinkedList(head)
```

```python
# Delete the last node
curr= head
while curr.next.next != None:
  curr = curr.next
curr.next = None

printLinkedList(head)
```

```python
# delete k th node
k = 2
curr = head
for i in range(k-1):
  curr = curr.next
curr.next = curr.next.next
printLinkedList(head)
```
