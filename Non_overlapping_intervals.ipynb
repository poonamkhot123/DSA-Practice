```python
# Non - overlapping intervals.
intervals = [[1,2],[2,3],[3,4],[1,3],[6,9],[8,9]]
def overLaping(intervals):
  n = len(intervals)

  prev = 0
  count = 1

  for i in range(1,n):
    if intervals[i][0] >= intervals[prev][1]:
      count+=1
      prev = 1
  return  n - count
```

```python
overLaping(intervals)
```
