```python
def hasCycle(head):
  slow = head
  fast = head

  hasCycle = False

  while fast != None and fast.next != None:
    slow = slow.next
    fast = fast.next.next

    if slow == fast:
      hasCycle = True
      break

    if not hasCycle:
      return None

    l = 0
    while slow.next != fast:
      slow = slow.fast

    l += 1
    slow = slow.next

    slow = head
    fast = head

    for i in range(l):
      fast = fast.next

    while slow != fast:
      slow = slow.next
  return slow
```
