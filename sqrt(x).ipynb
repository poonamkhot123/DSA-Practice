```python
def mySqrt(x):
  l = 1
  r = x

  while l<=r:
    mid = (l+r)//2
    midSquare = mid* mid

    if midSquare>x:
      r = mid-1
    else:
      ans = mid
      l = mid+1
  return ans
```

```python
x = 91
mySqrt(x)
```
