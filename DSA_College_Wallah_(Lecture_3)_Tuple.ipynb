```python
# Running sum of 1D array.
def runningSum(nums):
  n = len(nums)

  ans = []
  ans.append(nums[0])

  for i in range(1,n):
    x = ans[i-1] + nums[i]
    ans.append(x)
  return ans
```

```python
nums = [2,3,4,566,43,53,2]
runningSum(nums)
```

```python
# Remove duplicates from sorted array.
nums = [3,3,4,4,4,5,6,7,8,9]
def removeDuplicates(nums):
  n = len(nums)

  start = 0
  for i in range(1,n):
    # unique element
    if nums[i] != nums[start]:
      start+=1
      nums[start] = nums[i]

  return start+1

```

```python
removeDuplicates(nums)
```

```python
# Remove duplicates from sorted array II
def removeDuplicates(nums):
  n = len(nums)

  if n<=2:
    return n

  start = 1
  for i in range(2,n):
    # unique element
    if nums[i]!=nums[start-1]:
      start+=1
      nums[start] = nums[i]
  return start+1
```
