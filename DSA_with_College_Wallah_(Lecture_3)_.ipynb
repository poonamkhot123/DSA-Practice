```python
# sort array by parity
def sortArray(nums):
  n = len(nums)
  start = 0
  for i in range(n):
    if nums[i]%2==0:
      temp = nums[i]
      nums[i] = nums[start]
      nums[start] = temp
      start+=1
  return nums
```

```python
# Maximum subarray
def maxSubArray(nums):
  curr_sum = 0
  max_sum = nums[0]

  for i in range(len(nums)):
    curr_sum+=nums[i]
    if curr_sum>max_sum:
      max_sum = curr_sum
    if curr_sum<0:
      curr_sum = 0
  return max_sum
```

```python
# Best time to buy and sell the stock II
def maxProfit(prices):
  profit = 0
  n = len(prices)

  for i in range(n-1):
    if prices[i+1]>prices[i]:
      profit += prices[i+1]-prices[i]
  return profit
```

```python
# Richest customer wealth
def maximumWealth(accounts):
  ans = 0
  for account in accounts:
    ans = max(ans,sum(account))
  return ans
```

```python
# Defanging an IP Address
def IPAddr(address):
  ans = ""
  for i in address:
    if i !=".":
      ans+=i
    else:
      ans+="[.]"
  return ans
```

```python
IPAddr("1.1.1.1")
```

```python
# Valid Palindrome
def isAlphanumeric(s):
  x = ord(s)
  if 97<=x<=122 or 65<=x<=90 or 48<=x<=57:
    return True
  return False

def isPalindrome(s):
  s = s.lower()

  i = 0
  j = len(s)-1

  while i<j:
    if not isAlphanumeric(s[i]):
      i+=1
    elif not isAlphanumeric(s[j]):
      j-=1
    elif s[i]==s[j]:
      i+=1
      j-=1
    else:
      return False
  return True
```

```python
s = " A man, a plan, a canal: Panama"
isPalindrome(s)
```

```python
# Reverse a string
def reverseString(s):
  s_list = list(s)
  i = 0
  j = len(s_list) -1

  while i<j:
    temp = s_list[i]
    s_list[i] = s_list[j]
    s_list[j] = temp
    i+=1
    j-=1
  return "".join(s_list)
```

```python
s = "poonam khot"
reversed_s = reverseString(s)
print(reversed_s)
```

```python
def reverseWords(s):
  s = (s.strip()).split()
  s.reverse()
  return " ".join(s)
```

```python
s = "poonam khot is a data analyst"
reverseWords(s)
print(s)
```

```python
# length of last word.
def length(s):
  s = s.strip()
  n = len(s)
  i = -1

  while i>=(-1*n) and s[i]!=" ":
    i-=1
  i+=1
  i*=-1

  return i
```

```python
s = " poonam khot is a data analyst"
print(length(s))
```
